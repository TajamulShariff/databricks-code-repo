```python
#libraries to be imported for machine learning.
import pandas as pd
import numpy as np
```

```python
heart=pd.read_csv(r"C:\Users\admin\Desktop\BEPEC\dataset\clevelanda.csv")
```

```python
heart.head()
```

```python
heart.shape
```

```python
heart.info()
```

```python
heart['ca'].value_counts()
#there are 4 ? marks in the above columns, we need to remove them..
```

```python
heart['thal'].value_counts()

#there are two questions marks in the thal columns, we need to remove them..
```

```python
heart['ca']=pd.to_numeric(heart['ca'],errors='coerce')
```

```python
heart['thal']=pd.to_numeric(heart['thal'], errors='coerce')
```

```python
heart.info()
```

```python
heart.isnull().sum()
```

```python
#now we need to remove above null values..

heart.dropna(inplace=True)
```

```python
heart.info()
```

```python
heart.isnull().sum()
```

```python
heart.shape
```

```python
X=heart.iloc[:,:13]
```

```python
X.head()
```

```python
y=heart.iloc[:,13:]
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)
```

```python
X_train.shape
```

```python
X_train.head()
```

```python
y_train.head()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
```

```python
model.fit(X_train, y_train)  #Training Completed..
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score
```

```python
#compare both predicted and actual values

accuracy_score(y_test, y_pred)

#We got below accuracy using logistic regression..
```
